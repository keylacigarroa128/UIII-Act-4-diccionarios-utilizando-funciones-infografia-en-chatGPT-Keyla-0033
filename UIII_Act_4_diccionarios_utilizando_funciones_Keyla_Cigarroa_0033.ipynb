### Tema: Estilista

```python
# Diccionario principal para almacenar estilistas
estilistas = {}

def agregar_estilista():
    nombre = input("Nombre del estilista: ")
    especialidad = input("Especialidad: ")
    salon = input("Salón de belleza: ")
    agenda = input("Agenda semanal: ")
    tecnicas = input("Técnicas dominadas: ")
    marcas = input("Marcas aliadas: ")
    premios = input("Premios: ")

    estilistas[nombre] = {
        "especialidad": especialidad,
        "salon": salon,
        "agenda": agenda,
        "tecnicas": tecnicas,
        "marcas": marcas,
        "premios": premios
    }
    print(f"\n¡Estilista {nombre} agregado con éxito!")

def listar_estilistas():
    if not estilistas:
        print("\nNo hay estilistas registrados.")
        return

    # Encabezado de la tabla
    print("\n" + "="*130)
    print(f"{'Nombre':<15} | {'Especialidad':<15} | {'Salón':<15} | {'Agenda':<15} | {'Técnicas':<20} | {'Marcas':<15} | {'Premios':<15}")
    print("-" * 130)

    # Uso de for key, value (clave valor) para llenar la tabla
    for nombre, info in estilistas.items():
        print(f"{nombre:<15} | "
              f"{info['especialidad']:<15} | "
              f"{info['salon']:<15} | "
              f"{info['agenda']:<15} | "
              f"{info['tecnicas']:<20} | "
              f"{info['marcas']:<15} | "
              f"{info['premios']:<15}")
    print("="*130)

def buscar_estilista():
    nombre = input("Nombre del estilista a buscar: ")
    if nombre in estilistas:
        info = estilistas[nombre]
        print(f"\nResultado para {nombre}:")
        for clave, valor in info.items():
            print(f"- {clave.capitalize()}: {valor}")
    else:
        print("Estilista no encontrado.")

def borrar_estilista():
    nombre = input("Nombre del estilista a eliminar: ")
    if nombre in estilistas:
        del estilistas[nombre]
        print(f"Registro de {nombre} eliminado correctamente.")
    else:
        print("No se encontró el estilista.")

def actualizar_estilista():
    nombre = input("Nombre del estilista a actualizar: ")
    if nombre in estilistas:
        print(f"Modificando datos de {nombre}...")
        estilistas[nombre]['especialidad'] = input("Nueva Especialidad: ")
        estilistas[nombre]['salon'] = input("Nuevo Salón: ")
        estilistas[nombre]['agenda'] = input("Nueva Agenda: ")
        estilistas[nombre]['tecnicas'] = input("Nuevas Técnicas: ")
        estilistas[nombre]['marcas'] = input("Nuevas Marcas: ")
        estilistas[nombre]['premios'] = input("Nuevos Premios: ")
        print("Actualización completada.")
    else:
        print("El estilista no existe.")

# Menú interactivo
while True:
    print("\n--- MENÚ DE GESTIÓN (SALÓN DE BELLEZA) ---")
    print("1. Agregar estilista")
    print("2. Listar estilistas (Ver Tabla)")
    print("3. Buscar estilista")
    print("4. Borrar estilista")
    print("5. Actualizar estilista")
    print("6. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        agregar_estilista()
    elif opcion == "2":
        listar_estilistas()
    elif opcion == "3":
        buscar_estilista()
    elif opcion == "4":
        borrar_estilista()
    elif opcion == "5":
        actualizar_estilista()
    elif opcion == "6":
        print("Cerrando sistema. ¡Buen día!")
        break
    else:
        print("Opción inválida.")
```
