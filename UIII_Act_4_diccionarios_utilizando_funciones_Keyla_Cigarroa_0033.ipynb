### Tema: Estilista

```python
# Diccionario principal
estilistas = {}

def agregar_estilista():
    nombre = input("Nombre: ")
    especialidad = input("Esp. (Color/Corte): ")
    salon = input("Salón: ")
    agenda = input("Agenda: ")
    tecnicas = input("Técnicas: ")
    marcas = input("Marcas: ")
    premios = input("Premios: ")

    estilistas[nombre] = {
        "Esp": especialidad,
        "Salón": salon,
        "Agnd": agenda,
        "Técn": tecnicas,
        "Marc": marcas,
        "Prem": premios
    }

def listar_estilistas():
    if not estilistas:
        print("\nLista vacía.")
        return

    # Tabla compacta (Ancho total reducido)
    print("\n" + "-"*85)
    print(f"{'Nombre':<10} | {'Esp':<8} | {'Salón':<10} | {'Agnd':<8} | {'Técn':<12} | {'Marc':<10} | {'Prem':<10}")
    print("-" * 85)

    # Uso de for clave, valor (key, value)
    for nombre, d in estilistas.items():
        print(f"{nombre[:10]:<10} | "
              f"{d['Esp'][:8]:<8} | "
              f"{d['Salón'][:10]:<10} | "
              f"{d['Agnd'][:8]:<8} | "
              f"{d['Técn'][:12]:<12} | "
              f"{d['Marc'][:10]:<10} | "
              f"{d['Prem'][:10]:<10}")
    print("-" * 85)

def buscar_estilista():
    nombre = input("Nombre a buscar: ")
    if nombre in estilistas:
        for k, v in estilistas[nombre].items():
            print(f"{k}: {v}")
    else:
        print("No encontrado.")

def borrar_estilista():
    nombre = input("Nombre a borrar: ")
    if nombre in estilistas:
        del estilistas[nombre]
        print("Eliminado.")

def actualizar_estilista():
    nombre = input("Nombre a actualizar: ")
    if nombre in estilistas:
        estilistas[nombre]['Esp'] = input("Nueva Esp: ")
        estilistas[nombre]['Salón'] = input("Nuevo Salón: ")
        print("Actualizado.")

# --- INFOGRAFÍA DE DATOS DEL PROYECTO ---
# Estudiante: Keyla Cigarroa
# Especialidad: Inteligencia Artificial
# Plantel: CBTIS No. 128
# Submódulo 2: Elabora proyectos de programación lógica
# Docente: Eliseo Nava

while True:
    print("\n1.Add 2.List 3.Search 4.Del 5.Update 6.Exit")
    opcion = input("Opción: ")

    if opcion == "1":
        agregar_estilista()
    elif opcion == "2":
        listar_estilistas()
    elif opcion == "3":
        buscar_estilista()
    elif opcion == "4":
        borrar_estilista()
    elif opcion == "5":
        actualizar_estilista()
    elif opcion == "6":
        break
```
